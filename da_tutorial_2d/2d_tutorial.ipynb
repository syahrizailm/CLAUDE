```python
import numpy as np
import pandas as pd
import plotly.express as px
# pd.set_option('display.max_rows', None)
# pd.set_option('display.max_columns', None)
import matplotlib.pyplot as plt
from matplotlib import rc
import plotly.graph_objects as go
from scipy.optimize import curve_fit
```

```python
def truemodel_target1(xt):
    return 2.0 - xt[0] + 2.0*xt[0]**2 - 3.0*xt[0]**3 + xt[1] + xt[0]*xt[1]

def truemodel_target2(xt):
    return truemodel_target1(xt) - 1.0 + 0.5*xt[0] - 0.25*xt[1]

def truemodel(xt):
    return np.column_stack((2.0 - xt[:,0] + 2.0*xt[:,0]**2 - 3.0*xt[:,0]**3 + xt[:,1] + xt[:,0]*xt[:,1],
                             2.0 - xt[:,0] + 2.0*xt[:,0]**2 - 3.0*xt[:,0]**3 + xt[:,1] + xt[:,0]*xt[:,1] - 1.0 + 0.5*xt[:,0] - 0.25*xt[:,1]))
```

```python
X_1d = np.linspace(0.0, 1.0, 100)
Y_1d = np.linspace(0.0, 1.0, 100)
X, Y = np.meshgrid(X_1d, Y_1d)
coords = np.column_stack((X.ravel(), Y.ravel()))

z_true = truemodel(coords)
```

```python
target1 = z_true[:,0].reshape(100,100)
target2 = z_true[:,1].reshape(100,100)
```

```python
true_target1 = go.Surface(z=target1, colorscale=['rgb( 99, 110, 250)','rgb( 99, 110, 250)'], x=X_1d, y=Y_1d, opacity=0.5, showscale=False)
true_target2 = go.Surface(z=target2, colorscale=['rgb(239, 101,  72)','rgb(239, 101,  72)'], x=X_1d, y=Y_1d, opacity=0.5, showscale=False)
```

```python
fig = go.Figure(data=[true_target1, true_target2])
fig.update_layout(width=1000,height=1000)
fig.update_layout(scene_aspectmode='cube')
fig.update_layout(
    scene=dict(
        xaxis=dict(title='descriptor1'),
        yaxis=dict(title='descriptor2'),
        zaxis=dict(title='target')
    )
)
fig.update_layout(
    title={
        'text': r'toy model',
        'x': 0.5,
        'y': 0.9,
        'xanchor': 'center',
        'yanchor': 'top',
        'font': dict(size=30)
    }
)
fig.update_layout(legend=dict(font=dict(size=20)))

fig.update_layout(
    legend=go.layout.Legend(
        itemsizing='constant'
    )
)

fig.show()
```
